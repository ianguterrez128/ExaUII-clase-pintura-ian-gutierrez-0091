```python
class Pintura:
    def __init__(self, nombreObra, artista, tecnica):
        self.nombreObra = nombreObra
        self.artista = artista
        self.tecnica = tecnica

    def leerDatos(self):
        self.nombreObra = input("Ingrese el nombre de la obra: ")
        self.artista = input("Ingrese el nombre del artista: ")
        self.tecnica = input("Ingrese la técnica utilizada: ")
        return self.nombreObra, self.artista, self.tecnica

    def mostrarDatos(self):
        print("Nombre de la Obra:", self.nombreObra)
        print("Artista:", self.artista)
        print("Técnica:", self.tecnica)
        return self.nombreObra, self.artista, self.tecnica

# Fin de clase Pintura

# --- Registro del primer objeto ---
print("Cuadro número 1")
cuadro1 = Pintura("", "", "")  # Creando objeto cuadro 1 con valores iniciales
cuadro1.leerDatos()
cuadro1.mostrarDatos()

print("-*-*-*-*-*-*-*-*-*-*-*")

# --- Registro del segundo objeto ---
print("Cuadro número 2")
cuadro2 = Pintura("", "", "")  # Creando objeto cuadro 2 con valores iniciales
cuadro2.leerDatos()
cuadro2.mostrarDatos()

print("-*-*-*-*-*-*-*-*-*-*-*")
print("Software de Control de pintura finalizado")
print("Software desarrollado por Ian Gutierrez NL = 22")
```
